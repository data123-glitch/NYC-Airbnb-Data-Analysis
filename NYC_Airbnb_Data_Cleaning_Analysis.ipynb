## 3. Data Cleaning

The dataset was cleaned by handling missing values, removing invalid records, correcting data types, and creating additional fields for analysis.

### 3.1 Handling Invalid Prices

Listings with a price of $0 were identified as invalid for price analysis and removed from the dataset.

```python
(df["price"] == 0).sum()
```

```python
df = df[df["price"] > 0]
```

```python
(df["price"] == 0).sum()
```

```python
df.dtypes
```

### 3.2 Converting Date Data Type

The `last_review` column was converted from an object data type to a datetime format for accurate date handling.

```python
df["last_review"] = pd.to_datetime(df["last_review"])
```

```python
df["last_review"].dtype
```

### 3.3 Handling Missing Text Values

Missing values in `name` and `host_name` were replaced with "Unknown" instead of removing the affected listings.

```python
df["name"] = df["name"].fillna("Unknown")
df["host_name"] = df["host_name"].fillna("Unknown")
```

```python
df[["name", "host_name"]].isnull().sum()
```

### 3.4 Handling Missing Review Data

Missing values in `reviews_per_month` were replaced with 0 because these listings had no recorded monthly review activity.

```python
df["reviews_per_month"] = df["reviews_per_month"].fillna(0)
```

```python
df["reviews_per_month"].isnull().sum()
```

## 4. Outlier Detection

The Interquartile Range (IQR) method was used to identify unusually high Airbnb prices. Outliers were flagged instead of being removed because high-priced listings can represent genuine properties.

### 4.1 IQR Method for Price

The first quartile (Q1), third quartile (Q3), and IQR were calculated to determine the upper price boundary.

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
(df["price"] > upper_bound).sum()
```

### 4.2 Flagging Price Outliers

Listings above the calculated upper price boundary were flagged as outliers while keeping the original price data unchanged.

```python
df["price_status"] = np.where(
    df["price"] > upper_bound,
    "Outlier",
    "Normal"
)
```

## Key Findings

### Data Cleaning
- The original dataset contained 48,895 Airbnb listings.
- 11 listings with a price of $0 were removed.
- The cleaned dataset contains 48,884 listings.
- Missing values in `name` and `host_name` were replaced with "Unknown".
- Missing values in `reviews_per_month` were replaced with 0.
- `last_review` was converted from object to datetime format.
- No duplicate rows were found.
- Price outliers were identified using the IQR method and flagged rather than removed.

### Exploratory Data Analysis
- Entire home/apt listings have the highest average price at approximately $211.81.
- Manhattan has the largest number of listings with 21,660 listings.
- Williamsburg has the highest number of listings with 3,919 listings.
- Manhattan has the largest number of listings with 21,660 listings.
- Shared rooms have the highest average availability at approximately 161.87 days.
- Private rooms have the highest average number of reviews at approximately 24.10.
- Sonder (NYC) has the highest number of listings among the hosts analyzed, with 327 listings.
- Approximately 79.4% of listings have reviews, while 20.6% have no review history.
- The correlation between price and number of reviews is very weak and negative (-0.0479).

### Conclusion

The analysis shows that Airbnb pricing varies considerably by room type and location.
Entire homes and apartments are generally more expensive than private and shared rooms,
while Manhattan has the highest average listing price. The analysis also shows that
listing price and review count have almost no meaningful linear relationship.
